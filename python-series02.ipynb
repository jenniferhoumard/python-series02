## 2. Sinusodial Equation 


$s(t)$ = Asin($2\pi$ $ft$ + $\phi$) + $D$

$s(t)$ = Asin($\omega$ $t$ + $\phi$) + $D$

- $A$ = amplitude
- $f$ = frequency in [Hz]
- $t$ = time in seconds [s]
- $\phi$ = phase shift in radians (horizontal shift)
- $D$ = vertical shift
- $\omega$ = angular frequency [rad/s]

```python
# 1. Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
t = np.linspace(0, 1, 1000) # linspace() creates evenly spaced numbers (0 = starting value, 1 = ending value, 1000 = total values)

# 2. Create a numpy array called s that contains the values of a sine function of t (s is for signal)
# Since t contains 1000 values, Python calculates the sine for every one of those 1000 time values.
s = np.sin(2 * np.pi * t) # simplified sinusoid with 1 oscillation per second. np.sine accesses the sine function.
c = np.cos(2 * np.pi * t)

# 3. Plot the cosine curve with the sine curve

# fig = the whole figure, ax1 = the first axis, which will use the left-hand y-axis
fig, ax1 = plt.subplots() #plt.subplots() creates a Matplotlib figure and an axis.

# a) Plot sine on the first y-axis
line1, = ax1.plot(t, s, color="blue", label="Sine") # plot t on shared x axis and s on first y axis and store it in variable line 1

# b) Add title and labels to the first axis
ax1.set_title("Sinusoidal functions")

ax1.set_xlabel("Time (s)")
ax1.set_ylabel("Sine Amplitude")

# c) Add grid
ax1.grid()

# d) Create a second y-axis on the right-hand side
ax2 = ax1.twinx() # the cosine x axis will be shared with the sine x axis

# e) Plot cosine on the second y-axis
line2, = ax2.plot(t, c, color="orange", label="Cosine") # plot t on the x axis and c on the second y axis and store it in a variable line2

# f) Add label to the second y-axis
ax2.set_ylabel("Cosine Amplitude")

# g) Add a legend containing both sine and cosine
ax1.legend([line1, line2], ["Sine", "Cosine"], loc="lower left") # label line1 and line2 as Sine and Cosine

# h) Save as PDF
plt.savefig("sine_cosine.pdf")

# Display the figure
plt.show()
```

### c) Final version

A. Create a function called sinusoid that :

- takes as input all parameters that you need to define the sinusoid

- returns a numpy array called x

B. Use the function sinusoid to create two numpy arrays s1 and s2 with the two different frequencies

C. Plot s1 and s2 as a function of t (as illustrated in the graph below).

Parameters:

$s(t)$ = Asin($2\pi$ $ft$ + $\phi$) + $D$

- $t$ = time
- $A$ = amplitude
- $f$ = frequency
- $\phi$ = phase
- D = offset

```python
# 1. Define a function called sinusoid
def sinusoid(time, amplitude, frequency, phase, offset): #include all the parameters needed

# Create a doc string that describes purpose of the function,the inputs, and what it returns.
    """
    Generate a sinusoidal signal.

    Parameters
    ----------
    time : numpy.ndarray
        Time values in seconds.
    amplitude : float
        Amplitude of the sinusoid.
    frequency : float
        Frequency in Hz.
    phase : float
        Phase shift in radians.
    offset : float
        Vertical offset of the sinusoid.

    Returns
    -------
    x : numpy.ndarray
        Generated sinusoidal signal.
    """
# the function will return the following curve 
    x = amplitude * np.sin(2 * np.pi * frequency * time + phase) + offset 
    return x
```

```python
# 2. Create two numpy arrays s1 and s2 with two different frequencies 

# Create a common time base so both signals can be compared over the same 1-second interval
t = np.linspace(0, 1, 1000) #linspace() creates evenly spaced numbers (0 = starting value, 1 = ending value, 1000 = total values) 

#Create two variables s1 and s2 with identical parameters except for frequency allowing the effect of frequency on the sinusoid to be compared
s1 = sinusoid(t, amplitude=1, frequency=1, phase=0, offset=0) #s1 stores the first signal, with a frequency of 1 Hz
s2 = sinusoid(t, amplitude=1, frequency=2, phase=0, offset=0) #s2 stores the second signal, with a frequency of 2 Hz

# 3.Plot both signals on the same figure to compare their frequencies visually
plt.plot(t, s1, color="blue", label="Sinus 1 [1 Hz]")
plt.plot(t, s2, color="orange", label="Sinus 2 [2 Hz]")

plt.title("Sinusoidal functions with Different Frequencies")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid()
plt.legend()

plt.show()
```

### a) Even or odd: 

Goal: Define what an even number is, create numbers 0–10, print whether each is even or odd, then improve the code quality step by step.

What is an even number, from a mathematical point of view?

- An even number is an integer that can be divided by 2 with no remainder.

- From a mathematical point of view, an even number is any integer that can be written as:

- $n = 2k$

- *where $k$ is also an integer.

- If dividing the number by 2 leaves a remainder of 1, the number is odd.

```python
# Create a series of integer numbers from 0 to 10
numbers = range(0, 11) # range() generates a sequence of integers.

# Go through each number in the series one at a time
for number in numbers: # for starts a loop. number is a temporary variable representing the current number during each loop. # in → tells Python to take values from numbers.
    
    # If number divided by 2 has a remainder of 0, the number is even.
    if number % 2 == 0: # % calculates the remainder after division. 
        # We divide by 2 because we are testing whether the number is even.
        # == → comparison operator meaning "is equal to".
        # 0 → the remainder we expect for an even number.
        print(number, "is even") # number → displays the actual current number.

    # If the remainder is not 0, the number is odd.
    else:
        print(number, "is odd") 
```

### b) Multiple or not:

Goal: Define what a multiple is, test numbers 0–20 for multiples of 3 and 5, then generalize it using functions.

What is a multiple of a number m, from a mathematical point of view?

- A number ($n$) is a multiple of another number ($m$) if ($n$) can be written as:

- $n$ = $m$ × $k$

- where $k$ is an integer.

- In other words, ($n$) is a multiple of ($m$) if dividing ($n$) by ($m$) gives no remainder.

```python
# 1. For each number, print if it is a multiple of 3 or not

# Create a series of integers from 0 to 20
numbers = range(0, 21)

# Go through each number in the series one at a time
for number in numbers:

    # Check if dividing the number by 3 leaves a remainder of 0
    if number % 3 == 0:
        print(number, "is a multiple of 3")

    # If there is a remainder, the number is not a multiple of 3
    else:
        print(number, "is not a multiple of 3")
```
