## 2. Sinusodial Equation 


$s(t)$ = Asin($2\pi$ $ft$ + $\phi$) + $D$

$s(t)$ = Asin($\omega$ $t$ + $\phi$) + $D$

- $A$ = amplitude
- $f$ = frequency in [Hz]
- $t$ = time in seconds [s]
- $\phi$ = phase shift in radians (horizontal shift)
- $D$ = vertical shift
- $\omega$ = angular frequency [rad/s]

```python
#import libraries

import numpy as np # for numerical analysis
import matplotlib.pyplot as plt # for graph and visuals
```

```python
# Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
t = np.linspace(0, 1, 1000) #linspace() creates evenly spaced numbers (0 = starting value, 1 = ending value, 1000 = total values) 

# Create a numpy array called s that contains the values of a sine function of t (s is for signal)
s = np.sin(2 * np.pi * t) #uses the simplified sinusoid 
```

```python
# Plot s as a function of t

plt.plot(t, s)

plt.title("Sinusoidal function")

plt.xlabel("Time (s)")
plt.ylabel("Signal")
plt.grid()

plt.savefig("sine_wave.png")
plt.show()
```
